```python
import pandas as pd
import sys
sys.path.append('../src')
from features import *

# 1. On charge le dataset
df = pd.read_csv('../data/archive (2)/new_data_urls.csv')

# 2. On prend un échantillon de 5 000 lignes pour aller vite
df_test = df.head(5000).copy()

# 3. On calcule les caractéristiques
df_test['longueur'] = df_test['url'].apply(url_length)
df_test['nb_tirets'] = df_test['url'].apply(count_hyphens)
df_test['a_https'] = df_test['url'].apply(has_https)
df_test['a_ip'] = df_test['url'].apply(has_ip_address)

# 4. On regarde la MOYENNE des caractéristiques pour les sites sains (1) et les phishing (0)
print("=== MOYENNES POUR LES SITES SAINS (1) ===")
print(df_test[df_test['status'] == 1][['longueur', 'nb_tirets', 'a_https', 'a_ip']].mean())

print("\n=== MOYENNES POUR LES SITES PHISHING (0) ===")
print(df_test[df_test['status'] == 0][['longueur', 'nb_tirets', 'a_https', 'a_ip']].mean())
```
