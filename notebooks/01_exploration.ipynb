```python
import pandas as pd

# On indique le sous-dossier 'archive (2)' dans le chemin
df = pd.read_csv('../data/archive (2)/new_data_urls.csv')

# Afficher les 5 premières lignes
df.head()
```

```python
# 1. Vérifier s'il y a des valeurs manquantes
print("Valeurs manquantes :")
print(df.isnull().sum())

# 2. Vérifier s'il y a des doublons (des URL qui apparaissent 2 fois)
print("\nNombre de doublons :", df.duplicated().sum())

# 3. Vérifier l'équilibre des classes (0 = phishing, 1 = sain)
print("\nRépartition (0 = Phishing, 1 = Sain) :")
print(df['status'].value_counts())

# 4. Afficher 3 exemples de sites sains et 3 de phishing
print("\n--- Exemples de sites sains (1) ---")
print(df[df['status'] == 1]['url'].head(3).tolist())

print("\n--- Exemples de phishing (0) ---")
print(df[df['status'] == 0]['url'].head(3).tolist())
```
